```python
# Two Pointer 
nums = [1,2,4,6,8,10]
target =10
left = 0
right = len(nums)-1

while left < right :
    current_sum = nums[left]+nums[right]

    if current_sum == target :
        print(nums[left],nums[right])
        break
    elif current_sum<target:
        left+=1
    else:
        right-=1


```

```python
#Kth Largest Element
nums = [3,2,1,5,6,4]

for _ in range(2):
  largest = nums[0]

  for i in nums:
        if i > largest:
            largest = i
            
            
  nums.remove(largest)

print("2nd Largest =", largest)
```

```python
#Palindrome :

num = 121
original = num
rev = 0

while num > 0 :

   digit = num%10 
   rev = rev * 10 + digit
   num //= 10 

if original  ==  rev :
   print ("is plaindrome")
else :
  print ("not an plaindrome")
  
```

```python
#Remove Duplicates Array 

arr = [1,2,2,3,4,4,5]

seen = set()

result=[]

for num in arr:
    if num not in seen:
        seen.add(num)
        result.append(num)
print(result)
```

```python
# Rotate Array Left By Postion

def reverse (arr, start, end):
    while start < end:
        arr[start],arr[end] = arr[end], arr[start]
        start+=1
        end-=1

def leftarray(arr,k):
    n = len(arr)
    k %=n 
   

    reverse(arr, 0, k - 1)
    print(arr)
    reverse(arr, k, n - 1)
    print(arr)
    reverse(arr, 0, n - 1)
    print(arr)

    return arr

arr = [1, 2, 3, 4, 5]
print(leftarray(arr, 2))
```

```python
#Right Rotate Array 

def reverse (arr,start,end):
    while start < end:
        arr[start],arr[end] = arr[end],arr[start]
        start+=1
        end-=1

def right_array(arr,k):
    n = len(arr)
    k%=n

    reverse(arr, 0, n - 1)
    reverse(arr, 0, k - 1)
    reverse(arr, k, n - 1)

    return arr
    

arr = [1, 2, 3, 4, 5]
print(right_array(arr, 2))
```
