```python
# Two Pointer 
nums = [1,2,4,6,8,10]
target =10
left = 0
right = len(nums)-1

while left < right :
    current_sum = nums[left]+nums[right]

    if current_sum == target :
        print(nums[left],nums[right])
        break
    elif current_sum<target:
        left+=1
    else:
        right-=1


```

```python
#Kth Largest Element
nums = [3,2,1,5,6,4]

for _ in range(2):
  largest = nums[0]

  for i in nums:
        if i > largest:
            largest = i
            
            
  nums.remove(largest)

print("2nd Largest =", largest)
```
